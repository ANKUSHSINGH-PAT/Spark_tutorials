```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('Basics').getOrCreate()
```

```python
df=spark.read.csv('loan_data.csv',header=True,inferSchema=True)
```

```python
df.show()
```

```python
df.filter(df['fico']>700).show()
```

```python
df_renamed.summary().show()
```
